uploading the data set into colab

```python
from google.colab import files
uploaded = files.upload()
print(uploaded)
```

```python
# Output the uploaded file names and sizes
for fn in uploaded.keys():
    print(f'User uploaded file "{fn}" with length {len(uploaded[fn])} bytes')

# Now you can read the uploaded file with pandas
import pandas as pd
data = pd.read_csv(next(iter(uploaded)))

# Display the first few rows
data.head()
```

creating a comparison chart

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data for comparison (Petrol Tax vs Petrol Consumption)
labels = data.index
petrol_tax = data['Petrol_tax']
petrol_consumption = data['Petrol_Consumption']

x = np.arange(len(labels))

width = 0.35

fig, ax = plt.subplots(figsize=(12,6))
ax.bar(x - width/2, petrol_tax, width, label='Petrol Tax')
ax.bar(x + width/2, petrol_consumption, width, label='Petrol Consumption')

ax.set_xlabel('Index')
ax.set_ylabel('Values')
ax.set_title('Petrol Tax vs Petrol Consumption')
ax.legend()

plt.show()
```

Bar Chart - Average Petrol Consumption by Petrol Tax

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("w1_petrol_consumption-220925-152435.csv")

plt.figure(figsize=(8,6))
sns.barplot(x='Petrol_tax', y='Petrol_Consumption', data=df)
plt.title('Average Petrol Consumption by Petrol Tax')
plt.show()
```

Pie Chart - Distribution of Petrol Tax Counts

```python
plt.figure(figsize=(7,7))
df['Petrol_tax'].value_counts().plot.pie(autopct='%1.1f%%', startangle=140)
plt.title('Petrol Tax Distribution')
plt.ylabel('')
plt.show()
```

Histogram - Distribution of Average Income

```python
plt.figure(figsize=(8,6))
plt.hist(df['Average_income'], bins=15, color='skyblue', edgecolor='black')
plt.title('Histogram of Average Income')
plt.xlabel('Average Income')
plt.ylabel('Frequency')
plt.show()
```

Box Plot - Petrol Consumption by Petrol Tax

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='Petrol_tax', y='Petrol_Consumption', data=df)
plt.title('Boxplot of Petrol Consumption by Petrol Tax')
plt.show()
```

Violin Plot - Distribution of Petrol Consumption for each Petrol Tax

```python
plt.figure(figsize=(8,6))
sns.violinplot(x='Petrol_tax', y='Petrol_Consumption', data=df)
plt.title('Violin Plot of Petrol Consumption by Petrol Tax')
plt.show()
```

Scatter Plot - Relationship between Average Income and Petrol Consumption

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='Average_income', y='Petrol_Consumption', data=df)
plt.title('Scatter Plot: Average Income vs Petrol Consumption')
plt.show()
```

Bubble Plot - Size of Points based on Paved Highways

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Average_income'], df['Petrol_Consumption'], s=df['Paved_Highways']/10, alpha=0.5)
plt.title('Bubble Plot: Income vs Petrol Consumption (Size by Paved Highways)')
plt.xlabel('Average Income')
plt.ylabel('Petrol Consumption')
plt.show()
```

Pair Plot - Pairwise Relationships in the Dataset

```python
sns.pairplot(df)
plt.suptitle('Pair Plot of Dataset', y=1.02)
plt.show()
```

Heatmap - Correlation Matrix

```python
plt.figure(figsize=(8,6))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap of Correlation Matrix')
plt.show()
```
